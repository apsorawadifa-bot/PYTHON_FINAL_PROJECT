```python
cols={'social_media_hours','relationship_status','internet_quality','coffee_consumption_per_day','assignment_score','AI_tool_usage_hours'}
```

```python
df.drop(columns=cols,inplace=True,errors='ignore')
df.head()
```

```python
df.rename(columns={"GPA": "gpa",}, inplace=True)
df.head()
```

```python
cols=['age','gpa','mental_stress_level','class_attendance_percent','study_hours_per_day','sleep_hours','screen_time_hours','gaming_hours','exercise_hours_per_week','part_time_job','extracurricular_hours_per_week',]
plt.figure(figsize=(16,12))
for i, col in enumerate(cols,1):
    plt.subplot(3, 4, i)
    sns.boxplot(y=df[col],width=0.5)
    plt.title(f'Boxplot of {col}',fontsize=12)
    plt.ylabel(' ')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.subplots_adjust(hspace=0.5, wspace=0.4)
plt.tight_layout()
plt.show()  
```
