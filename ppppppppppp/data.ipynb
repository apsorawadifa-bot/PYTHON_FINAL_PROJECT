```python
cols=['age','gpa','mental_stress_level',"university_year",'class_attendance_percent','study_hours_per_day','sleep_hours','screen_time_hours','gaming_hours','exercise_hours_per_week','extracurricular_hours_per_week',]
plt.figure(figsize=(16,12))
for i, col in enumerate(cols,1):
    plt.subplot(3, 4, i)
    sns.boxplot(y=df_clean[col],width=0.5)
    plt.title(f'Boxplot of {col}',fontsize=12)
    plt.ylabel(' ')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.subplots_adjust(hspace=0.5, wspace=0.4)
plt.tight_layout()
plt.show()  
```

```python
Q1 = df[cols].quantile(0.25)
Q3 = df[cols].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df_clean[cols] = np.where(df_clean[cols] < lower,lower, df_clean[cols])
df_clean[cols] = np.where(df_clean[cols] > upper,upper, df_clean[cols])

```

```python
df_clean.shape
```

```python
for col in ['gpa', 'screen_time_hours']:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean[col] = df_clean[col].clip(lower, upper)
```

```python
for col in ['gpa', 'screen_time_hours']:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[(df_clean[col] < lower) |(df_clean[col] > upper)]

    print(col, len(outliers))
```
